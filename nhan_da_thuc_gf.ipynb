# Phép nhân 2 đa thức trong trường GF(2^4) và GF(2^8)

```python
def nhan_da_thuc(a, b, so_bit, m_x):
    gioi_han = 1 << so_bit
    ket_qua = 0
    for _ in range(so_bit):
        if b & 1:
            ket_qua = ket_qua ^ a
        bit_cao_nhat = a & (1 << (so_bit - 1))
        a = a << 1
        if bit_cao_nhat:
            a = a ^ m_x
        a = a & (gioi_han - 1)
        b = b >> 1
    return ket_qua
```

## Phép nhân trong GF(2^4)

GF(2^4) dùng đa thức tối giản $x^4+x+1$.

```python
def gf2_4(a, b):
    m_x = 0b10011  # x^4 + x + 1
    return nhan_da_thuc(a, b, 4, m_x)
```

```python
# Vi du: a = x^2 + x + 1; b = x^3 + x + 1
a = 0b0111
b = 0b1011
ket_qua = gf2_4(a, b)

print("a        =", format(a, "04b"))
print("b        =", format(b, "04b"))
print("kết quả =", format(ket_qua, "04b"))
```

## Phép nhân trong GF(2^8)

GF(2⁸) dùng đa thức tối giản $x^8+x^4+x^3+x+1$

```python
def gf2_8(a, b):
    m_x = 0b100011011  # x^8 + x^4 + x^3 + x + 1
    return nhan_da_thuc(a, b, 8, m_x)
```

```python

a = 0b00100110
b = 0b10011110
ket_qua = gf2_8(a, b)

print("a        =", format(a, "08b"))
print("b        =", format(b, "08b"))
print("kết quả =", format(ket_qua, "08b"))
```
